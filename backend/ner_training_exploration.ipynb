```python
def process_receipt(filename):

    with open(os.path.join(entities_folder, filename), "r") as f:
        entities = json.load(f)

    with open(os.path.join(box_folder, filename), "r") as f:
        box = f.readlines()

    clean_text = []

    for text in box:
        parts = text.split(",")
        just_text = parts[-1]
        cleaned = just_text.replace("\n", "")
        clean_text.append(cleaned)

    full_text = " ".join(clean_text)
    labels_to_use = ["company", "date", "total"]
    found_entities = []


    for label in labels_to_use:
        value = entities[label]
        start = full_text.find(value)

        if start != -1:
            end = start + len(value)
            found_entities.append((start, end , label.upper()))

    return (full_text, {"entities": found_entities})
            
        

    
        
    
```

```python
example = process_receipt(filename[0])
print(example)
```

```python
process_receipt(filename[1])
```

```python
training_data = []

for f in filename:
    example = process_receipt(f)
    training_data.append(example)

print(len(training_data))
```

```python
full_matches = 0
partial_matches = 0


for text, annotation in training_data:
    num_found = len(annotation["entities"])
    if  num_found == 3:
        full_matches +=1
    else:
        partial_matches +=1
        
print("Full matches:", full_matches)
print("Partial matches:", partial_matches)
```

```python
import random

random.shuffle(training_data)

split_point = int(len(training_data) * 0.9)

train_set = training_data[:split_point]
test_set = training_data[split_point:]


print(len(train_set), len(test_set))
```

```python
import spacy

nlp = spacy.blank("en")
```

```python
ner = nlp.add_pipe("ner")
```

```python
labels = ["COMPANY", "DATE", "TOTAL"]

for label in labels:
    ner.add_label(label)
```

```python
from spacy.training import Example


text, annotation = train_set[0]
doc = nlp.make_doc(text)
example = Example.from_dict(doc, annotation)

print(example)
```

```python
nlp.initialize()


for epoch in range(20):
    random.shuffle(train_set)
    losses = {}


    for text, annotation in train_set:
        doc = nlp.make_doc(text)
        example = Example.from_dict(doc, annotation)
        nlp.update([example], losses=losses)

    print(f"Epoch: {epoch + 1}, Losses: {losses}")
```

```python
text, annotations = test_set[0]
doc = nlp(text)

for ent in doc.ents:
    print(ent.text, ":", ent.label_)
```

```python
for text, annotations in test_set[:5]:
    doc = nlp(text)
    print("PREDICTED:")
    for ent in doc.ents:
        print(f"  {ent.label_}: {ent.text}")
    print("---")
```
